```python
asia = df[df['origin'] == 'Asia'][["vehicle_weight", "model_year"]].head(7)
```

```python
X = asia.to_numpy()
```

```python
import numpy as np
```

```python
XTX = np.dot(X.T, X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```
